```python
import random

def partition(arr, left, right):
    """
    Partitions the array around a random pivot element.
    Elements smaller than pivot go left, larger go right.
    """
    pivot_index = random.randint(left, right)
    arr[pivot_index], arr[right] = arr[right], arr[pivot_index]

    pivot_value = arr[right]
    store_index = left

    for i in range(left, right):
        if arr[i] < pivot_value:
            arr[i], arr[store_index] = arr[store_index], arr[i]
            store_index += 1

    arr[store_index], arr[right] = arr[right], arr[store_index]
    return store_index

def quickselect(arr, left, right, k_smallest):
    """
    Finds the k-th smallest element (0-indexed) in arr[left...right].
    """
    if left == right:
        return arr[left]

    pivot_index = partition(arr, left, right)

    if k_smallest == pivot_index:
        return arr[pivot_index]
    elif k_smallest < pivot_index:
        return quickselect(arr, left, pivot_index - 1, k_smallest)
    else:
        # Fixed: Corrected positional arguments to match (arr, left, right, k_smallest)
        return quickselect(arr, pivot_index + 1, right, k_smallest)

def find_kth_smallest(arr, k):
    """
    Wrapper function converting 1-based index 'k' to 0-based indexing.
    """
    if k < 1 or k > len(arr):
        raise ValueError("k must be between 1 and the length of the array")
    return quickselect(arr.copy(), 0, len(arr) - 1, k - 1)

# ==========================================
# Testing Framework
# ==========================================
if __name__ == "__main__":
    # Generate 1000 random integers between -5000 and 5000
    random.seed(42)
    n = 1000
    test_array = [random.randint(-5000, 5000) for _ in range(n)]

    # Standard Python sorted values to verify correctness
    validation_sorted = sorted(test_array)

    # 1. Test for k = 1 (Minimum element)
    k_min = 1
    result_min = find_kth_smallest(test_array, k_min)
    expected_min = validation_sorted[0]

    # 2. Test for k = n (Maximum element)
    k_max = n
    result_max = find_kth_smallest(test_array, k_max)
    expected_max = validation_sorted[-1]

    # 3. Test for k = n // 2 (Median element)
    k_med = n // 2
    result_med = find_kth_smallest(test_array, k_med)
    expected_med = validation_sorted[k_med - 1]

    # Display Results
    print(f"Array Size: {n}\n" + "="*45)
    print(f"k = 1 (Min):     QuickSelect = {result_min:<6} | Expected = {expected_min}")
    print(f"k = n (Max):     QuickSelect = {result_max:<6} | Expected = {expected_max}")
    print(f"k = n//2 (Med):  QuickSelect = {result_med:<6} | Expected = {expected_med}")

    # Final Validation Check
    assert result_min == expected_min, "Min verification failed!"
    assert result_max == expected_max, "Max verification failed!"
    assert result_med == expected_med, "Median verification failed!"
    print("="*45 + "\nAll QuickSelect checks passed successfully! ✅")
```
