```python
import heapq

def find_fastest_route(graph, start, destination):
    # Initialize distances with infinity and start node with 0
    distances = {node: float('inf') for node in graph}
    distances[start] = 0

    # Track predecessors to reconstruct the path
    predecessors = {node: None for node in graph}

    # Priority queue stores tuples of (travel_time, node)
    priority_queue = [(0, start)]

    while priority_queue:
        current_time, current_node = heapq.heappop(priority_queue)

        # Stop early if we reached destination with the optimal path
        if current_node == destination:
            break

        # Skip if a faster path to this node was already processed
        if current_time > distances[current_node]:
            continue

        # Check neighboring nodes
        for neighbor, travel_time in graph[current_node].items():
            new_time = current_time + travel_time

            # If a shorter route is found, update records
            if new_time < distances[neighbor]:
                distances[neighbor] = new_time
                predecessors[neighbor] = current_node
                heapq.heappush(priority_queue, (new_time, neighbor))

    # Reconstruct the optimal path from destination back to start
    path = []
    current = destination
    while current is not None:
        path.append(current)
        current = predecessors[current]
    path.reverse()

    return distances[destination], path

# Define the 5-node city network map (one-way roads with times in minutes)
city_network = {
    0: {1: 4, 2: 2},          # From Airport (0)
    1: {3: 5},                # From Node 1
    2: {1: 1, 3: 8, 4: 10},   # From Node 2
    3: {4: 2},                # From Node 3
    4: {}                     # Central Station (4) has no outgoing roads
}

# Run the algorithm
time_taken, fastest_route = find_fastest_route(city_network, start=0, destination=4)

print(f"Fastest Route: {' -> '.join(map(str, fastest_route))}")
print(f"Total Travel Time: {time_taken} minutes")
```
