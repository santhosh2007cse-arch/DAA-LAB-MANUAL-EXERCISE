```python
import time

# -----------------------------
# Top-Down (Memoization)
# -----------------------------
def matrix_chain_memo(dims):
    n = len(dims) - 1
    memo = [[None] * (n + 1) for _ in range(n + 1)]

    def solve(i, j):
        if i == j:
            return 0

        if memo[i][j] is not None:
            return memo[i][j]

        memo[i][j] = float('inf')

        for k in range(i, j):
            cost = (
                solve(i, k)
                + solve(k + 1, j)
                + dims[i - 1] * dims[k] * dims[j]
            )

            if cost < memo[i][j]:
                memo[i][j] = cost

        return memo[i][j]

    return solve(1, n)


# -----------------------------
# Bottom-Up Dynamic Programming
# -----------------------------
def matrix_chain_bottom_up(dims):
    n = len(dims) - 1

    m = [[0] * (n + 1) for _ in range(n + 1)]

    for length in range(2, n + 1):
        for i in range(1, n - length + 2):
            j = i + length - 1
            m[i][j] = float('inf')

            for k in range(i, j):
                cost = (
                    m[i][k]
                    + m[k + 1][j]
                    + dims[i - 1] * dims[k] * dims[j]
                )

                if cost < m[i][j]:
                    m[i][j] = cost

    return m[1][n]


# -----------------------------
# Main Program
# -----------------------------
# Example chain of 6 matrices
# A1 = 30x35
# A2 = 35x15
# A3 = 15x5
# A4 = 5x10
# A5 = 10x20
# A6 = 20x25

dims = [30, 35, 15, 5, 10, 20, 25]

# Top-Down
start = time.perf_counter()
memo_cost = matrix_chain_memo(dims)
memo_time = time.perf_counter() - start

# Bottom-Up
start = time.perf_counter()
dp_cost = matrix_chain_bottom_up(dims)
dp_time = time.perf_counter() - start

# Results
print("Matrix Chain Multiplication Comparison")
print("--------------------------------------")
print("Dimensions:", dims)

print("\nTop-Down (Memoization)")
print("Minimum Cost:", memo_cost)
print("Execution Time: {:.6f} seconds".format(memo_time))

print("\nBottom-Up (Dynamic Programming)")
print("Minimum Cost:", dp_cost)
print("Execution Time: {:.6f} seconds".format(dp_time))

print("\nVerification")
if memo_cost == dp_cost:
    print("✔ Both methods give the SAME minimum cost.")
else:
    print("✘ The results are different.")
```
