```python
def matrix_chain_order(dims):
    """
    Matrix Chain Multiplication using Dynamic Programming

    Time Complexity: O(n^3)
    Space Complexity: O(n^2)
    """

    n = len(dims) - 1

    # DP table for minimum multiplication cost
    m = [[0] * (n + 1) for _ in range(n + 1)]

    # Table to store optimal split position
    s = [[0] * (n + 1) for _ in range(n + 1)]

    # Chain length
    for l in range(2, n + 1):
        for i in range(1, n - l + 2):
            j = i + l - 1
            m[i][j] = float('inf')

            for k in range(i, j):
                cost = (
                    m[i][k]
                    + m[k + 1][j]
                    + dims[i - 1] * dims[k] * dims[j]
                )

                if cost < m[i][j]:
                    m[i][j] = cost
                    s[i][j] = k

    return m, s


def print_optimal_parens(s, i, j):
    if i == j:
        matrices = ["T", "R", "S", "V", "P"]
        return matrices[i - 1]

    k = s[i][j]

    left = print_optimal_parens(s, i, k)
    right = print_optimal_parens(s, k + 1, j)

    return f"({left} x {right})"


def print_dp_table(m, n):
    print("\nDP Cost Table (m[i][j]):")

    print(f'{"":>6}', end="")
    for j in range(1, n + 1):
        print(f"M{j:>10}", end="")
    print()

    for i in range(1, n + 1):
        print(f"M{i:<5}", end="")
        for j in range(1, n + 1):
            if j < i:
                print(f'{"---":>11}', end="")
            else:
                print(f"{m[i][j]:>11}", end="")
        print()


# ----------------------------------------------------
# Matrix dimensions
# T(4x4), R(4x4), S(4x4), V(4x10000), P(10000x1)
# ----------------------------------------------------

dims = [4, 4, 4, 4, 10000, 1]

matrix_names = ["T", "R", "S", "V", "P"]

print("Matrix Dimensions:")
for i in range(len(matrix_names)):
    print(f"{matrix_names[i]} : {dims[i]} x {dims[i+1]}")

# Compute optimal order
m, s = matrix_chain_order(dims)

n = len(matrix_names)

print("\nMinimum Scalar Multiplications =", m[1][n])
print("Optimal Parenthesization =", print_optimal_parens(s, 1, n))

# Display DP table
print_dp_table(m, n)
```
