```python
def find_top_two_min_max(arr, low, high):
    """
    Finds (min1, min2, max1, max2) simultaneously
    using a Divide and Conquer approach.
    """
    # Counter for tracker comparisons
    comp_count = 0

    # Base Case 1: Only 1 element
    if low == high:
        # min1, min2, max1, max2
        # (Using float('inf') and float('-inf') to handle missing second elements)
        return (arr[low], float('inf'), arr[low], float('-inf')), 0

    # Base Case 2: Only 2 elements
    if high == low + 1:
        comp_count += 1
        if arr[low] < arr[high]:
            return (arr[low], arr[high], arr[high], arr[low]), comp_count
        else:
            return (arr[high], arr[low], arr[low], arr[high]), comp_count

    # Divide Step
    mid = (low + high) // 2
    (left_min1, left_min2, left_max1, left_max2), left_comp = find_top_two_min_max(arr, low, mid)
    (right_min1, right_min2, right_max1, right_max2), right_comp = find_top_two_min_max(arr, mid + 1, high)

    comp_count += left_comp + right_comp

    # Conquer / Combine Step
    # 1. Combine Minimums
    comp_count += 1
    if left_min1 < right_min1:
        min1 = left_min1
        comp_count += 1
        min2 = min(left_min2, right_min1)
    else:
        min1 = right_min1
        comp_count += 1
        min2 = min(left_min1, right_min2)

    # 2. Combine Maximums
    comp_count += 1
    if left_max1 > right_max1:
        max1 = left_max1
        comp_count += 1
        max2 = max(left_max2, right_max1)
    else:
        max1 = right_max1
        comp_count += 1
        max2 = max(left_max1, right_max2)

    return (min1, min2, max1, max2), comp_count


# --- Verification Example ---
if __name__ == "__main__":
    test_arr = [12, 13, 1, 10, 34, 1, 2, 45]
    print(f"Input Array: {test_arr}\n")

    (m1, m2, mx1, mx2), total_comps = find_top_two_min_max(test_arr, 0, len(test_arr) - 1)

    print(f"Minimum: {m1}, Second Minimum: {m2}")
    print(f"Maximum: {mx1}, Second Maximum: {mx2}")
    print(f"Total element comparisons: {total_comps}")
```
